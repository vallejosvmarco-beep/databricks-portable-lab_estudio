```python
def pasar_tabla_voluen(tabla, tipo):

    ruta_salida = (
        "/Volumes/workspace/alunizaje/datos_oficiales/"
        f"exportaciones/{tipo}"
    )
    df = spark.table(tabla)

    (
        df.write
        .mode("overwrite")
        .parquet(ruta_salida)
    )

    print(ruta_salida)
```
